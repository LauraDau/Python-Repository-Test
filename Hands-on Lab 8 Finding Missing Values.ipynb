### 3. Finding Missing Values
##### Task 2: Identify missing values for all columns.


```python
df.isnull().sum()
```

```python
with pd.option_context('display.max_rows', None,'display.max_columns',None):
    print(df.isnull().sum())
```

##### Task 3: Visualize missing values using a heatmap (Using seaborn library).



```python
# select the data from the dataset that is just missing values
missing_values=df.isnull().sum()


# Set a threshold to filter out columns with very few missing values
missing_threshold = 45000  # Adjust as needed
missing_values = df.isnull().sum()

# Select only columns with more than the threshold of missing values
cols_with_missing = missing_values[missing_values > missing_threshold].index

# Plot the heatmap with only these columns
plt.figure(figsize=(15, 10))
sns.heatmap(df[cols_with_missing].isnull(), cbar=False, cmap='Blues', yticklabels=False)
plt.title('Heatmap of Missing Values (Filtered)', fontsize=16)
plt.show()


```

##### Task 4: Count the number of missing rows for a specific column (e.g., `Employment`).

### 4. Imputing Missing Values
##### Task 5: Identify the most frequent (majority) value in a specific column (e.g., `Employment`).


```python
df['Employment'].value_counts().to_frame()
```

##### Task 6: Impute missing values in the `Employment` column with the most frequent value.



```python
most_freq=df['Employment'].value_counts().idxmax()
df_EdLevel_fill = df[['Employment']].fillna(most_freq)
```
