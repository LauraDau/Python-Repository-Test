### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
# --- Section 6: Encoding Categorical Data ---

# 6.1 Identify and convert categorical variables into numerical format

# Step 1: Split multiple responses in Employment column
df['Employment'] = df['Employment'].str.split(';')

# Step 2: Perform multi-hot encoding for Employment
employment_dummies = df['Employment'].str.join('|').str.get_dummies()

# Step 3: Join encoded Employment columns back to the main DataFrame
df_encoded = df.join(employment_dummies)

# Step 4: One-hot encode other categorical columns (e.g., Country, OpSysPersonal use)
df_encoded = pd.get_dummies(
    df_encoded,
    columns=['Country', 'OpSysPersonal use'],
    prefix=['Country', 'OS'],
    drop_first=True   # optional – avoids dummy variable trap
)


```

```python

# Step 4: One-hot encode other categorical columns (e.g., Country, OpSysPersonal use)
df_encoded = pd.get_dummies(
    df_encoded,
    columns=['Country', 'OpSysPersonal use'],
    prefix=['Country', 'OS'],
    drop_first=True   # optional – avoids dummy variable trap
)

# --- Section 6.2 Verify the Encoded Columns ---

# Step 1: Check encoded columns information
df_encoded.info()

# Step 2: View first few rows
df_encoded.head()

# Step 3: Check for missing values (optional)
df_encoded.isnull().sum()

# Step 4: Fill any remaining missing values with 0 (optional)
df_encoded.fillna(0, inplace=True)
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
df.isnull().sum()
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
mean=df['ConvertedCompYearly'].value_counts().mean()
```

```python
import numpy as np
```

```python
df['ConvertedCompYearly'].replace(np.nan, mean, inplace=True)
print ("columns have been filled")
```

```python
df.isnull().sum()
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
most_freq=df['RemoteWork'].value_counts().idxmax()
df['RemoteWork'].replace(np.nan, most_freq, inplace=True)
print ("columns have been filled")
```

```python
df.isnull().sum()
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
df['ConvertedCompYearly_MinMax'] = (
    (df['ConvertedCompYearly'] - df['ConvertedCompYearly'].min()) /
    (df['ConvertedCompYearly'].max() - df['ConvertedCompYearly'].min())
)
```
